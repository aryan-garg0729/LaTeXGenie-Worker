```python
from flask import Flask, request, send_file
from pyngrok import ngrok
import os, zipfile

app = Flask(__name__)
@app.route('/',methods=['GET'])
def home():
  return "hello"

@app.route('/convert', methods=['POST'])
def convert_pdf():
    uploaded_file = request.files['file']

    input_path = '/content/input.pdf'
    output_dir = '/content/output'
    zip_path = '/content/output.zip'

    uploaded_file.save(input_path)

    os.system(f'rm -rf {output_dir} {zip_path}')
    # if conda+pip magic-pdf
    # os.system(f'magic-pdf -p {input_path} -o {output_dir}')
    os.system(f'python /content/LaTeXGenie-Core/magic_pdf/tools/cli.py -p {input_path} -o {output_dir}')

    with zipfile.ZipFile(zip_path, 'w') as zipf:
        for root, _, files in os.walk(output_dir):
            for file in files:
                zipf.write(os.path.join(root, file), os.path.relpath(os.path.join(root, file), output_dir))

    return send_file(zip_path, as_attachment=True)

# Start ngrok tunnel
public_url = ngrok.connect(80)
print("Public URL:", public_url)  # put this public url in parser

app.run(host="0.0.0.0", port=80)
```
